```python
from gmaps_fetcher import search_by_name, fetch_reviews, fetch_photo_urls
from tqdm import tqdm

SAMPLE = 5 # set to None to run all rows

def enrich(row):
    place = search_by_name(row['HotelName'], region='台北')
    if not place or not place.get('place_id'):
        return pd.Series({
            'chinese_name': None,
            'gmaps_url': None,
            'review_summary': None,
            'top_pics': None,
        })

    pid = place['place_id']
    reviews = fetch_reviews(pid, sort='most_relevant')
    summary = ' | '.join(r['text'] for r in reviews[:3] if r.get('text'))
    pics = fetch_photo_urls(pid, max_photos=5)

    return pd.Series({
        'place_id': place.get('place_id'),
        'chinese_name': place.get('name'),
        'gmaps_url': place.get('google_maps_url'),
        'review_summary': summary or None,
        'top_pics': pics or None,
    })



target = df.head(SAMPLE) if SAMPLE else df
tqdm.pandas(desc='Enriching')
enriched = target.progress_apply(enrich, axis=1)

df_enriched = target.join(enriched)
df_enriched
```

```python
import os
from google import genai
from gmaps_fetcher import fetch_reviews

gemini = genai.Client(api_key=os.environ['GEMINI_API_KEY'])

def summarize_reviews(reviews: list[dict]) -> str | None:
    texts = [r['text'] for r in reviews if r.get('text')]
    if not texts:
        return None

    joined = '\n---\n'.join(texts)
    prompt = (
        "以下是某間飯店的 Google 評論。請用 2-3 句繁體中文總結整體住客體驗，"
        "包含優點與缺點，避免逐條重複，僅輸出摘要本身：\n\n"
        f"{joined}"
    )

    resp = gemini.models.generate_content(
        model='gemini-2.5-flash',
        contents=prompt,
    )
    return resp.text.strip() if resp.text else None
```

```python
from tqdm import tqdm
tqdm.pandas(desc='Summarising')

def llm_summary(row):
    place = search_by_name(row['HotelName'], region='台北')
    if not place or not place.get('place_id'):
        return None
    reviews = fetch_reviews(place['place_id'], sort='most_relevant')
    return summarize_reviews(reviews)

df_enriched['review_summary'] = df_enriched.progress_apply(llm_summary, axis=1)
df_enriched[['HotelName', 'chinese_name', 'review_summary']].head()
```
